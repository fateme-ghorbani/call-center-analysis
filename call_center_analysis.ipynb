## Loading Dataset and Initial Inspection

# Data Cleaning

### Categorical & Numeric Inspection

```python
categorical_cols = [
    "call_reason",  
    "queue",
    "channel",
    "status",
    "first_contact_resolution"
]

for col in categorical_cols:
    print(f"\n--- {col} ---")
    print(df[col].value_counts(dropna=False))
```

```python
numeric_cols = [
    "wait_seconds",
    "handle_seconds",
    "hold_seconds",
    "after_call_work_seconds",
    "csat_score",
    "transfer_count"
]

for col in numeric_cols:
    print(f"\n--- {col} ---")
    print(df[col].describe())
```

### Negative & Zero Values

```python
for col in [
    "wait_seconds",
    "handle_seconds",
    "hold_seconds",
    "after_call_work_seconds",
]:
    print(f"\n--- {col} ---")
    print("Negative values:", (df[col] < 0).sum())
    print("Zero values:", (df[col] == 0).sum())
```

### CSAT Values

```python
print(df["csat_score"].value_counts(dropna=False).sort_index())
```

### Inspect CSAT = 6

```python
df[df["csat_score"] == 6]
```

### CSAT Missingness by Status

```python
pd.crosstab(
    df["status"],
    df["csat_score"].isna()
)
```

### Create Clean Handle Time

```python
df["handle_seconds_clean"] = df["handle_seconds"].where(
    df["handle_seconds"] >= 0,
    np.nan
)
```

### Investigate Duplicate Call IDs

```python
df[df.duplicated("call_id", keep=False)].sort_values("call_id")
```

### Remove Exact Duplicates

```python
df = df.drop_duplicates()
print("Shape after removing exact duplicates:", df.shape)
print("Exact duplicates:", df.duplicated().sum())
```

### Investigate inconsistent duplicate


```python
df[df["customer_phone"].astype(str).str.startswith("+98")]
```

### Remove the inconsistent duplicate

```python
df = df[df["call_id"] != "C0500"]

print("Shape after removing inconsistent duplicate:", df.shape)
```

### Validate Duplicate Removal

```python
print("Unique Call IDs:", df["call_id"].nunique())
print("Exact Duplicates:", df.duplicated().sum())
```

### Missing Agent IDs

```python
df[df["agent_id"].isna()]
```

### Status vs Handle Time

```python
pd.crosstab(
    df["status"],
    df["handle_seconds_clean"].notna())
```

### Find Logical Anomalies


```python
df[
    ((df["status"] == "Abandoned") & df["handle_seconds_clean"].notna()) |
    ((df["status"] == "Answered") & df["handle_seconds_clean"].isna())
]
```

### Exclude Handle Time for Abandoned Calls

```python
df.loc[
    df["status"] == "Abandoned",
     "handle_seconds_clean"
     ]= np.nan
```

### Hold vs Handle Consistency

```python
(df["hold_seconds"] > df["handle_seconds_clean"]).sum()
```

### Convert Date & Time

```python
df["call_date"] = pd.to_datetime(df["call_date"])

df["start_time"] = pd.to_datetime(
    df["start_time"],
    format="%H:%M:%S"
)
```

### Validate Conversion

```python

print(df["call_date"].isna().sum())
print(df["start_time"].isna().sum())
```

### Flag potentially unusual CSAT values

```python
df["csat_invalid"] = df["csat_score"] > 5

print("Potentially unusual CSAT values:")
print(
    df.loc[
        df["csat_invalid"],
        ["call_id", "csat_score"]
    ]
)
```

CSAT values above 5 were flagged for investigation. Since the dataset documentation does not define a strict CSAT scale, these values were retained in the analysis.

### Extract Hour

```python
df["hour"] = df["start_time"].dt.hour

df["hour"].value_counts().sort_index()
```

### Define Shifts

```python
conditions = [
    df["hour"].between(8, 12),
    df["hour"].between(14, 19)
]

choices = ["Morning", "Evening"]

df["shift"] = np.select(conditions, choices, default="Unknown")
```

### Validate Shifts

```python
df["shift"].value_counts()
```

### Save Cleaned Dataset

```python
df.to_csv(
    "data/call_center_cleaned.csv",
    index=False
)

print("Cleaned dataset saved successfully.")
```

# KPI Analysis

### Call counts by status

```python
status_counts = df["status"].value_counts()
total_calls = len(df)
answered_calls = (df["status"] == "Answered").sum()
transferred_calls = (df["status"] == "Transferred").sum()
abandoned_calls = (df["status"] == "Abandoned").sum()

print("Total Calls:", total_calls)
print("Answered:", answered_calls)
print("Transferred:", transferred_calls)
print("Abandoned:", abandoned_calls)
```

### Answer Rate

```python
handled_calls = answered_calls + transferred_calls

answer_rate = (handled_calls / total_calls) * 100

print("Answer Rate:", round(answer_rate, 2), "%")
```

### Abandon Rate

```python
total_calls = len(df)
abandoned_calls = (df["status"] == "Abandoned").sum()

abandon_rate = (abandoned_calls / total_calls) * 100

print("Total Calls:", total_calls)
print("Abandoned Calls:", abandoned_calls)
print("Abandon Rate:", round(abandon_rate, 2), "%")
```

### Average Wait Time (AWT)


```python
awt = df["wait_seconds"].mean()

print("Average Wait Time (AWT):", round(awt, 2), "seconds")
```

### Average Handle Time (AHT)

```python
aht = df["handle_seconds_clean"].mean()

print("Average Handle Time (AHT):", round(aht, 2), "seconds")
```

### Customer Satisfaction (CSAT)

```python

csat_mean = df["csat_score"].mean()

print("Average CSAT:", round(csat_mean, 2))
```

### FCR distribution

```python
df["first_contact_resolution"].value_counts(dropna=False)
```

### First Contact Resolution (FCR)

```python
fcr_valid = df["first_contact_resolution"].dropna()

fcr = (fcr_valid == "Yes").mean() * 100

print("FCR:", round(fcr, 2), "%")
```

### Transfer Rate


```python
transferred_calls = (df["status"] == "Transferred").sum()
total_calls = len(df)

transfer_rate = (transferred_calls / total_calls) * 100

print("Transfer Rate:", round(transfer_rate, 2), "%")
```

### Average After Call Work (ACW)

```python
acw = df["after_call_work_seconds"].mean()

print("Average ACW:", round(acw, 2), "seconds")
```

# Team-level KPI Analysis

### KPI by Team

```python
team_kpis = []

for team, group in df.groupby("queue"):

    total_calls = len(group)

    handled_calls = group["status"].isin(
        ["Answered", "Transferred"]
    ).sum()

    abandoned_calls = (
        group["status"] == "Abandoned"
    ).sum()

    transferred_calls = (
        group["status"] == "Transferred"
    ).sum()

    fcr_valid = group["first_contact_resolution"].dropna()

    team_kpis.append({
        "Team": team,
        "Volume": total_calls,
        "Answer Rate (%)": round(
            handled_calls / total_calls * 100, 2
        ),
        "Abandon Rate (%)": round(
            abandoned_calls / total_calls * 100, 2
        ),
        "AWT (sec)": round(
            group["wait_seconds"].mean(), 2
        ),
        "AHT (sec)": round(
            group["handle_seconds_clean"].mean(), 2
        ),
        "CSAT": round(
            group["csat_score"].mean(), 2
        ),
        "FCR (%)": round(
            (fcr_valid == "Yes").mean() * 100, 2
        ),
        "Transfer Rate (%)": round(
            transferred_calls / total_calls * 100, 2
        )
    })

team_kpis_df = pd.DataFrame(team_kpis)

team_kpis_df
```

### Call Reason Distribution by Team

```python
team_reason = pd.crosstab(
    df["queue"],
    df["call_reason"]
)

team_reason
```

### AHT by Call Reason

```python
aht_by_reason = (
    df.groupby("call_reason")["handle_seconds_clean"]
      .mean()
      .round(2)
      .sort_values(ascending=False)
)

aht_by_reason
```

### AHT by Team and Call Reason

```python
aht_team_reason = (
    df.groupby(["queue", "call_reason"])["handle_seconds_clean"]
      .mean()
      .round(2)
      .unstack()
)

aht_team_reason
```

### Hold Time, ACW and Transfer Rate by Team

```python
team_operational = (
    df.groupby("queue")
      .agg(
          Hold_Time_sec=("hold_seconds", "mean"),
          ACW_sec=("after_call_work_seconds", "mean"),
          Transfer_Rate=(
              "status",
              lambda x: (x == "Transferred").mean() * 100
          )
      )
      .round(2)
)

team_operational
```

### AHT by Transfer Status

```python
df["is_transferred"] = df["status"] == "Transferred"

aht_by_transfer = (
    df.groupby("is_transferred")["handle_seconds_clean"]
      .mean()
      .round(2)
)

aht_by_transfer
```

### Transfer Rate by Team and Call Reason

```python
transfer_by_reason = (
    df.groupby(["queue", "call_reason"])["status"]
      .apply(lambda x: (x == "Transferred").mean() * 100)
      .round(2)
      .unstack()
)

transfer_by_reason
```

### AHT by Team and Transfer Status

```python
aht_transfer_team = (
    df.groupby(["queue", "is_transferred"])["handle_seconds_clean"]
      .mean()
      .round(2)
      .unstack()
)

aht_transfer_team
```

### Agent-level AHT Analysis

```python
agent_aht = (
    df[df["queue"] == "Support"]
    .groupby("agent_id")
    .agg(
        Calls=("call_id", "count"),
        AHT_sec=("handle_seconds_clean", "mean"),
        Transfer_Rate=(
            "status",
            lambda x: (x == "Transferred").mean() * 100
        )
    )
    .round(2)
    .sort_values("AHT_sec", ascending=False)
)

agent_aht
```

### A07 AHT by Call Reason

```python
df[
    (df["queue"] == "Support") &
    (df["agent_id"] == "A07")
].groupby("call_reason")["handle_seconds_clean"].agg(
    Calls="count",
    AHT_sec="mean"
).round(2)
```

### A07 Operational Metrics

```python
agent07_operational = (
    df[
        (df["queue"] == "Support") &
        (df["agent_id"] == "A07")
    ][
        ["handle_seconds_clean", "hold_seconds", "after_call_work_seconds"]
    ]
    .mean()
    .round(2)
)

agent07_operational
```
